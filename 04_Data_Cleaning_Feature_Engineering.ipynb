```python
import pandas as pd
import numpy as np

df = pd.read_csv("application_train.csv")

print(df.shape)
df.head()
```

```python
# Missing values check

missing = df.isnull().sum().sort_values(ascending=False)

missing_percent = (missing / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing_Count': missing,
    'Missing_%': missing_percent
})

missing_df.head(20)
```

```python
# Remove columns having more than 50% missing values

threshold = 50

cols_to_drop = missing_percent[missing_percent > threshold].index

df = df.drop(columns=cols_to_drop)

print("Dropped columns:", len(cols_to_drop))
print("New shape:", df.shape)
```

```python
# Fill remaining missing values

for col in df.columns:
    
    if df[col].dtype == 'object':
        # Text/category columns
        df[col] = df[col].fillna(df[col].mode()[0])
        
    else:
        # Numeric columns
        df[col] = df[col].fillna(df[col].median())

print("Remaining missing values:")
print(df.isnull().sum().sum())
```

```python
# Fill remaining missing values - fixed version

for col in df.columns:

    if df[col].dtype == 'object':
        df[col] = df[col].fillna(df[col].mode()[0])

    else:
        try:
            df[col] = df[col].fillna(df[col].median())
        except:
            df[col] = df[col].fillna(0)

print("Remaining missing values:")
print(df.isnull().sum().sum())
```

```python
# Feature Engineering

# Income per family member
df['INCOME_PER_PERSON'] = df['AMT_INCOME_TOTAL'] / (df['CNT_FAM_MEMBERS'] + 1)

# Credit to Income ratio
df['CREDIT_INCOME_RATIO'] = df['AMT_CREDIT'] / (df['AMT_INCOME_TOTAL'] + 1)

# Employment ratio
df['EMPLOYED_RATIO'] = df['DAYS_EMPLOYED'] / (df['DAYS_BIRTH'] + 1)

print(df[['INCOME_PER_PERSON',
          'CREDIT_INCOME_RATIO',
          'EMPLOYED_RATIO']].head())
```

```python
# Save cleaned dataset

df.to_csv("cleaned_credit_data.csv", index=False)

print("Dataset saved successfully!")
```
